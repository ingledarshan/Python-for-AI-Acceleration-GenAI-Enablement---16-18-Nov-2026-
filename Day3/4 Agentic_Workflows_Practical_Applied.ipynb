## **Problem Statement**

In many real-world scenarios, users seek contextual information about the current date and time as part of their natural language queries. For example:

* *“What is today’s date?”*
* *“Can you tell me the current time before I schedule a meeting?”*

While large language models (LLMs) can generate approximate or static answers, they are **not inherently aware of real-time system information**. As a result, they may provide outdated or fabricated responses when asked about the current date or time.

To ensure accuracy, an intelligent assistant must:

1. **Understand natural language queries referring to temporal context.**
2. **Automatically invoke tools that fetch real-time system date and time.**
3. **Integrate tool outputs into a conversational, human-friendly response.**

This functionality is crucial in **productivity tools, scheduling assistants, personal reminders, and business applications**, where real-time awareness significantly enhances user experience and reliability.

---

## **Objective**

Design and implement a **LangChain-powered agent** that integrates real-time utilities for date and time. The agent should:

* **Detect** when a user query involves current date or time.
* **Automatically call the appropriate tool** (`get_current_date` or `get_current_time`).
* **Leverage LangChain’s tool-calling framework** to handle reasoning and orchestration.
* **Maintain a reasoning trace** using `agent_scratchpad` for clarity.
* **Provide responses** in a conversational and context-aware format, ensuring outputs are both precise and user-friendly.

```python
import requests
from datetime import datetime, timezone
import ntplib

# What is ntplib?
# ntplib is a Python library used to interact with Network Time Protocol (NTP) servers. It allows you to retrieve the current time from NTP servers, which is useful for applications that require accurate timekeeping.
# This function fetches the current UTC time from an internet source using ntplib and falls back to Google's server time if ntplib fails.

def get_internet_utc():
    try:
        client = ntplib.NTPClient()
        response = client.request("pool.ntp.org")
        print("Hitting WorldTimeAPI, got data:", response)
        return datetime.fromtimestamp(response.tx_time, timezone.utc)
    except Exception:
        r = requests.head("https://google.com")
        server_time = r.headers["Date"]
        print("Hitting Google, got server time:", server_time)
        return datetime.strptime(server_time, "%a, %d %b %Y %H:%M:%S %Z").replace(tzinfo=timezone.utc)

utc_now = get_internet_utc()
print("Reliable UTC time:", utc_now)

# Take Home Task: Replace the current tools with nltplib from the above lines of code
```

```python
# Tool Definitions
import pendulum

@tool
def get_current_date_in_india():
    """Returns the current date in India (Asia/Kolkata timezone) in format YYYY-MM-DD."""
    now = pendulum.now("Asia/Kolkata")
    return f"Current date in India is {now.format('YYYY-MM-DD')}."

@tool
def get_current_time_in_india():
    """Returns the current time in India (Asia/Kolkata timezone) in format HH:mm:ss."""
    now = pendulum.now("Asia/Kolkata")
    return f"Current time in India is {now.format('HH:mm:ss')}."

tools = [get_current_date_in_india, get_current_time_in_india]
```

```python
agent = create_tool_calling_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
```

```python
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True # Set verbose=True to see the agent's reasoning and tool calls in the output
)
```

```python
# Load the requested secrets locally; prefer my.env and fall back to the project's .env.
import os
from pathlib import Path
from dotenv import load_dotenv

for env_file in (Path("my.env"), Path(".env")):
    if env_file.is_file():
        load_dotenv(env_file, override=False)

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
OPENWEATHERMAP_API_KEY = os.getenv("OPENWEATHERMAP_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
if not OPENAI_API_KEY:
    raise ValueError("OPENAI_API_KEY was not found in the environment, my.env, or .env.")
```

```python
# Load the policy PDF as one source-aware LangChain document per page.
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader

pdf_path = Path("docs/company_policy.pdf")
if not pdf_path.is_file():
    pdf_path = Path("company_policy.pdf")
if not pdf_path.is_file():
    raise FileNotFoundError("Put company_policy.pdf in docs/ or next to this notebook.")

documents = [page for page in PyPDFLoader(str(pdf_path)).load() if page.page_content.strip()]
if not documents:
    raise ValueError("No PDF text was extracted. Check the file or use OCR for scanned pages.")

print(f"Loaded {len(documents)} pages with text from {pdf_path.name}.")
print(documents[0].page_content[:250])
```

```python
# Split pages with the same baseline chunk size as notebook 3 and preserve page metadata.
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)
chunks = text_splitter.split_documents(documents)
if not chunks:
    raise ValueError("The PDF did not produce any non-empty text chunks.")

# Label only the teaching pages used by the stipend and probation examples.
topics_by_page = {4: "recruitment", 5: "recruitment", 6: "recruitment",
                  7: "probation", 8: "probation"}
tagged_chunks = []
for chunk in chunks:
    page_number = chunk.metadata["page"] + 1
    metadata = {
        **chunk.metadata,
        "source": pdf_path.name,
        "pdf_page": page_number,
        "topic": topics_by_page.get(page_number, "other"),
    }
    tagged_chunks.append(Document(page_content=chunk.page_content, metadata=metadata))

print(f"Created {len(tagged_chunks)} page-aware chunks.")
```

```python
# Build the same small, in-memory OpenAI vector index used in notebook 3.
from time import perf_counter
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=OPENAI_API_KEY,
    timeout=60,
    max_retries=2,
)
started = perf_counter()
vectorstore = InMemoryVectorStore.from_documents(
    documents=tagged_chunks,
    embedding=embeddings,
)
print(f"Indexed {len(tagged_chunks)} chunks in {perf_counter() - started:.2f}s.")
```

```python
# Keep retrieval, source formatting, and context limits in one reusable place.
def search_policy(query, k=3, topic=None):
    def matches_topic(document):
        return topic is None or document.metadata.get("topic") == topic

    return vectorstore.similarity_search_with_score(query, k=k, filter=matches_topic)


def format_hits(hits, max_chars=1400):
    blocks = []
    used_chars = 0
    for document, score in hits:
        page = document.metadata.get("pdf_page", document.metadata.get("page", -1) + 1)
        block = f"[p. {page}] {document.page_content.strip()}"
        added_chars = len(block) + (2 if blocks else 0)
        if used_chars + added_chars > max_chars:
            continue
        blocks.append(block)
        used_chars += added_chars
    return "\n\n".join(blocks)


# Inspect the same baseline question used in notebook 3 before introducing agent decisions.
baseline_question = "If I join this company as an intern, will I get any stipend?"
baseline_hits = search_policy(baseline_question, k=3)
baseline_context = format_hits(baseline_hits)
print(f"Retrieved {len(baseline_hits)} passages; assembled {len(baseline_context)} characters.")
for document, score in baseline_hits:
    print(f"PDF page {document.metadata['pdf_page']} | score {score:.3f} | "
          f"{' '.join(document.page_content.split())[:180]}")
```

## 1. Agent Fundamentals (Applied Perspective)

An **agent** combines an LLM with application logic and tools. The LLM proposes an action; the application validates and executes it, adds the result to state, and decides whether to continue. Tools do not run just because a model mentions them in ordinary text: the application must dispatch the structured tool call.

### Chain or agent?

| Prefer a chain when… | Consider an agent when… |
| --- | --- |
| Steps and order are known | The next tool depends on what the user asks or what a tool returns |
| You want predictable latency, cost, and testing | The benefit of adaptive tool choice justifies extra latency and failure modes |
| One retrieval and one answer are enough | The system must select, inspect, and possibly repeat a tool call |

**When agents add value:** choosing between a policy search and a page lookup, or retrying a permitted operation after a transient failure. **When they overcomplicate:** a single fixed search-answer flow, simple validation, or a task that can be handled by a direct function.

**Common failure patterns:** selecting the wrong tool; repeatedly calling a tool; using irrelevant or empty evidence; treating retrieved text as instructions; hiding a tool failure; and answering confidently after a failed lookup. The flow below bounds tool rounds, keeps evidence in state, and tells the model not to invent facts.


```python
# Define the fixed RAG baseline for comparison; defining the chain does not call the model.
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o-mini",
    api_key=OPENAI_API_KEY,
    temperature=0,
    timeout=60,
    max_retries=2,
)
NO_EVIDENCE = "I could not find this information in the provided SPIL policy document."
rag_prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer briefly using only the supplied SPIL policy passages. "
     f"If they do not support an answer, say exactly: {NO_EVIDENCE}"),
    ("human", "Policy passages:\n{context}\n\nQuestion: {question}"),
])
rag_chain = rag_prompt | llm | StrOutputParser()
print("The fixed RAG chain is ready; no chat completion was requested.")
```

## Chain or agent? Make the design decision first

The fixed RAG chain always retrieves the top passages and generates an answer. The **policy agent** can choose between policy search and page lookup. The **general tool agent** can also choose local date/time, current weather, and Tavily search when the request needs current external data. For a static question that needs none of those sources, it should make no tool call.

This illustrates the chain-vs-agent decision: extra tools are useful only when the request requires them. Otherwise, a fixed chain is simpler, more predictable, and usually cheaper.

**Tool examples:**

- **Data retrieval:** search the policy index or fetch a source page.
- **Internal service:** a narrowly scoped employee or order lookup, with authorization and timeouts enforced by its service wrapper.
- **External API:** a documented API operation with validated inputs, bounded retries, and a clear error result (OpenWeatherMap and Tavily examples follow).

The internal-service example is design guidance; this notebook does not connect to a private company service.

## 2. Agent Flow Design

Use LangChain's `create_tool_calling_agent` and `AgentExecutor` for the agent workflow:

1. Create a prompt containing the user request and an agent scratchpad.
2. Bind only the tools allowed for this run and let the model choose whether to request any.
3. Let `AgentExecutor` execute tool actions, append observations, and ask the model for its next response.
4. Keep execution bounded with `max_iterations`; return the framework's intermediate steps for inspection.

The application adds two narrow safeguards around the framework: a hard limit of two tool calls in a single model turn and a retry wrapper for transient tool failures. It does not hand-roll the agent loop.
